## Tips

1. Use plan mode before writing the code to keep the context window concise and clean. Frequently correcting for the wrong active explodes the context window and make model forget what its initial requirements
2. Dont be afraid to stop it half way. Check the thinking process and stop if it is going the wrong way. Use Esc to stop it.
3. Try to break down a complex problem yourself or let the model give you a plan before executing.
4. For a complicated request, it is worth to write a good prompt. Write your request, let the model help you to improve the prompt then run it. Also, add think a lot to trigger higher level of thinking mode.
